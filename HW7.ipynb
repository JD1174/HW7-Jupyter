# New York Times API Interface: Most Popular Articles

The objective of this notebook is to construct a robust Python interface that queries the New York Times Most Popular API, retrieves real-time JSON data regarding the most viewed articles over the last 7 days, and transforms it into a clean, structured pandas DataFrame ready for downstream analysis.

### API Configuration & Request Setup

We utilize the provided API credentials for authentication. We will query the `/viewed/7.json` endpoint to fetch the articles that have garnered the most views on NYTimes.com over the past week.

```python
import requests
import pandas as pd

# API Configuration
# Replace the placeholder below with your personal New York Times API Key
API_KEY = 'INSERT_YOUR_OWN_API_KEY_HERE'
BASE_URL = 'https://api.nytimes.com/svc/mostpopular/v2/viewed/7.json'

# Construct request parameters
params = {
    'api-key': API_KEY
}

print("=== Requesting data from NYT Most Popular API ===")


response = requests.get(BASE_URL, params=params)
if response.status_code == 200:
    data = response.json()
    print(f"Success! Retrieved {len(data.get('results', []))} articles.")
else:
    print(f"Error: Failed to fetch data. Status Code: {response.status_code}")
```

### Transforming JSON Payload to a Pandas DataFrame

With a successful API response, we will extract the core `results` list and read it into a pandas DataFrame. To ensure our dataset is highly readable and analytical, we will select and clean a subset of the most relevant columns, such as the title, section, author, publication date, and URL.

```python
# Extract and transform to DataFrame
if response.status_code == 200 and 'results' in data:
    articles_list = data['results']

    # Load JSON into DataFrame
    df_raw_articles = pd.DataFrame(articles_list)

    # Select key attributes for our structured table
    columns_to_keep = ['id', 'title', 'byline', 'section', 'published_date', 'type', 'url']
    df_articles = df_raw_articles[columns_to_keep].copy()

    # Rename columns for pristine presentation
    df_articles.columns = ['ID', 'Title', 'Byline', 'Section', 'Published Date', 'Type', 'URL']

    # Set ID as the index
    df_articles.set_index('ID', inplace=True)

    print("=== Most Popular NYT Articles (Last 7 Days) ===")
    display(df_articles.head(10))
else:
    print("No data available to construct DataFrame.")
```

### Analytical Summary & Discussion

1. **Data Completeness**: The DataFrame cleanly maps nested JSON structures into a highly accessible tabular format. Columns like `Byline` and `Section` allow for straightforward grouping and aggregation.
2. **Downstream Applications**: With this baseline structured DataFrame, we can easily perform text mining on article titles, analyze the "velocity" of publications by day of the week, or categorize trends across different editorial desks.
